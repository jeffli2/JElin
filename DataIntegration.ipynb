```python
import pandas as pd
import numpy as np

# Load the two datasets
print("Loading datasets...")
world_data = pd.read_csv("world-data-2023-(1)-csv-csv.csv")
happiness = pd.read_csv("World-happiness-report-2024-csv-csv.csv")

print(f"World Data shape: {world_data.shape}")
print(f"Happiness Report shape: {happiness.shape}")
```

```python
print("Cleaning World Data")

# Standardize country names to uppercase for consistency
world_data['Country'] = world_data['Country'].str.upper()

# Select and reorder columns from world data
world_data = world_data[[
    "Country", "Abbreviation", "Capital/Major City", "Population", 
    "Density\n(P/Km2)", "Land Area(Km2)", "GDP", "Minimum wage",
    "Currency-Code", "Tax revenue (%)", "Unemployment rate",
    "Life expectancy", "Birth Rate", "Fertility Rate",
    "Infant mortality", "Latitude", "Longitude"
]]

# Rename columns for clarity
world_data = world_data.rename(columns={
    "Density\n(P/Km2)": "Density (P/Km2)",
    "Land Area(Km2)": "Land Area (Km2)"
})

print("World Data columns selected:")
print(world_data.columns.tolist())
print("\nWorld Data head:")
print(world_data.head())
```

```python
print("\n=== Cleaning Happiness Report ===")

# Standardize country names to uppercase
happiness['Country name'] = happiness['Country name'].str.upper()

# Handle name mismatches
name_mapping = {
    'CZECHIA': 'CZECH REPUBLIC',
    'TURKIYE': 'TURKEY',
    'TAIWAN PROVINCE OF CHINA': 'TAIWAN',
    'CONGO (BRAZZAVILLE)': 'REPUBLIC OF THE CONGO',
    'CONGO (KINSHASA)': 'DEMOCRATIC REPUBLIC OF THE CONGO',
    'GAMBIA': 'THE GAMBIA'
}
happiness['Country name'] = happiness['Country name'].replace(name_mapping)

# Select and rename columns from happiness report
happiness = happiness[[
    "Country name", "Regional indicator", "Ladder score",
    "Log GDP per capita", "Social support", "Healthy life expectancy",
    "Freedom to make life choices", "Generosity",
    "Perceptions of corruption", "Dystopia + residual"
]]

happiness = happiness.rename(columns={
    "Country name": "Country",
    "Ladder score": "Happiness Score"
})

print("Happiness Report columns selected:")
print(happiness.columns.tolist())
print("\nHappiness Report head:")
print(happiness.head())
```

```python
print("\n=== Merging Datasets ===")

# Merge on Country name (inner join - only keep matching countries)
final_df = pd.merge(
    world_data,
    happiness,
    on='Country',
    how='inner'
)

print(f"Merged dataset shape: {final_df.shape}")
print(f"Successfully merged {len(final_df)} countries")

# Add a RecordID column (like in the inspections example)
final_df.index = range(1, len(final_df) + 1)
final_df['RecordID'] = final_df.index

# Reorder columns with RecordID first
final_df = final_df[[
    'RecordID', 'Country', 'Abbreviation', 'Regional indicator',
    'Capital/Major City', 'Population', 'Density (P/Km2)', 'Land Area (Km2)',
    'GDP', 'Minimum wage', 'Currency-Code', 'Happiness Score',
    'Log GDP per capita', 'Social support', 'Healthy life expectancy',
    'Freedom to make life choices', 'Generosity', 'Perceptions of corruption',
    'Dystopia + residual', 'Tax revenue (%)', 'Unemployment rate',
    'Life expectancy', 'Birth Rate', 'Fertility Rate', 'Infant mortality',
    'Latitude', 'Longitude'
]]

print("\n=== Final Integrated Dataset ===")
print(final_df.head(10))
print(f"\nTotal rows: {len(final_df)}")
print(f"Total columns: {len(final_df.columns)}")
```

```python
print("\n=== Saving Dataset ===")
final_df.to_csv('World_Happiness_Integrated.csv', index=False)
print("✓ Saved full dataset: World_Happiness_Integrated.csv")
```

```python
print("Reduced Dataset")
# Sample 75% of the data
final_reduced = final_df.sample(frac=0.75, random_state=477)
print(f"Reduced dataset shape: {final_reduced.shape}")
print(final_reduced.head())

final_reduced.to_csv('World_Happiness_Integrated_Reduced.csv', index=False)
print("✓ Saved reduced dataset: World_Happiness_Integrated_Reduced.csv")
```

```python
print("\n=== Data Quality ===")
print(f"Missing GDP values: {final_df['GDP'].isna().sum()}")
print(f"Missing Happiness scores: {final_df['Happiness Score'].isna().sum()}")
print(f"Missing Minimum wage: {final_df['Minimum wage'].isna().sum()}")
print(f"Missing Dystopia values: {final_df['Dystopia + residual'].isna().sum()}")

print("\n=== Countries Based on Region ===")
print(final_df['Regional indicator'].value_counts())

print("\n✓ Integration")
```
